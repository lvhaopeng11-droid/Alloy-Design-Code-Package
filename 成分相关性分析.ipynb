```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 读取数据集
file_path = '皮尔逊分析.xlsx'
data = pd.read_excel(file_path)
# 查看数据的前几行，确保数据读取成功
print(data.head())

# 计算相关性矩阵
correlation_matrix = data.corr()

# 获取特征名称
feature_names = data.columns

# 设置绘图样式
sns.set(style='white')

# 创建图形并设置分辨率
plt.figure(figsize=(10, 8), dpi=600)

# 使用imshow绘制相关性矩阵热图
plt.imshow(correlation_matrix, cmap='coolwarm', vmin=-1, vmax=1)

# 添加颜色条
plt.colorbar(label='Correlation Coefficient')

# 设置x和y轴的标签
plt.xticks(np.arange(len(feature_names)), feature_names, fontstyle='italic',rotation=0, fontweight='bold', fontsize=18, fontname='Times New Roman')
plt.yticks(np.arange(len(feature_names)), feature_names, fontstyle='italic', fontweight='bold', fontsize=18, fontname='Times New Roman')

# 获取当前坐标轴对象
ax = plt.gca()

# 为每个单元格添加黑色边框
for i in range(len(feature_names)):
    for j in range(len(feature_names)):
        # 绘制文本标签
        plt.text(j, i, '{:.2f}'.format(correlation_matrix.iloc[i, j]), ha='center', va='center', color='black', fontweight='bold', fontsize=18, fontname='Times New Roman')
        # 设置每个矩阵单元格的边框，颜色为黑色，粗细为1
        ax.add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1, fill=False, edgecolor='black', lw=1))

# 调整布局，确保标签不被遮挡
plt.tight_layout()

# 显示图形
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 读取数据集
file_path = '工作簿3.xlsx'
data = pd.read_excel(file_path)
# 查看数据的前几行，确保数据读取成功
print(data.head())

# 计算相关性矩阵
correlation_matrix = data.corr()

# 只保留上三角部分
correlation_matrix = np.triu(correlation_matrix)

# 获取特征名称
feature_names = data.columns

# 设置绘图样式
sns.set(style='white')

# 创建图形并设置分辨率
plt.figure(figsize=(10, 8), dpi=600)

# 使用imshow绘制相关性矩阵热图，只显示右上角部分
plt.imshow(correlation_matrix, cmap='coolwarm', vmin=-1, vmax=1)

# 添加颜色条
plt.colorbar(label='Correlation Coefficient')

# 设置x和y轴的标签
plt.xticks(np.arange(len(feature_names)), feature_names, fontstyle='italic', rotation=45, fontweight='bold', fontsize=18, fontname='Times New Roman')
plt.yticks(np.arange(len(feature_names)), feature_names, fontstyle='italic', fontweight='bold', fontsize=18, fontname='Times New Roman')

# 获取当前坐标轴对象
ax = plt.gca()

# 为每个单元格添加黑色边框
for i in range(len(feature_names)):
    for j in range(i, len(feature_names)):  # 只显示上三角部分
        if not np.isnan(correlation_matrix[i, j]):  # 忽略NaN值
            # 绘制文本标签
            plt.text(j, i, '{:.2f}'.format(correlation_matrix[i, j]), ha='center', va='center', color='black', fontweight='bold', fontsize=18, fontname='Times New Roman')
            # 设置每个矩阵单元格的边框，颜色为黑色，粗细为1
            ax.add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1, fill=False, edgecolor='black', lw=1))

# 调整布局，确保标签不被遮挡
plt.tight_layout()

# 显示图形
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. 读取数据
file_path = '工作簿3.xlsx'
data = pd.read_excel(file_path)
print(data.head())  # 检查数据是否正确读取

# 2. 计算相关性矩阵
correlation_matrix = data.corr()

# 3. 创建掩码，保留上三角，遮住下三角
mask = np.tril(np.ones_like(correlation_matrix, dtype=bool))

# 4. 设置画布大小与分辨率
plt.figure(figsize=(10, 8), dpi=600)

# 5. 绘制热图
#   - mask=mask 只显示上三角
#   - annot=True 在每个单元格显示数值
#   - fmt='.2f' 数值保留两位小数
#   - annot_kws 设置数字字体加粗、Times New Roman等
#   - cbar_kws 设置颜色条标签等
ax = sns.heatmap(
    correlation_matrix, 
    mask=mask, 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    annot=True, 
    fmt='.2f',
    annot_kws={
        "fontweight": "bold",
        "fontname": "Times New Roman",
        "fontsize": 14,
        "color": "black"
    },
    cbar_kws={"label": "Correlation Coefficient"}
)

# 6. 设置坐标轴字体
plt.xticks(
    rotation=0,  # 若想倾斜显示可改为 45
    fontweight='bold',
    fontname='Times New Roman',
    fontsize=14
)
plt.yticks(
    fontweight='bold',
    fontname='Times New Roman',
    fontsize=14
)

# 7. 获取并设置颜色条 (Colorbar) 字体
cbar = ax.collections[0].colorbar
# 颜色条刻度字体
cbar.ax.tick_params(labelsize=14)
for tick_label in cbar.ax.yaxis.get_ticklabels():
    tick_label.set_fontweight("bold")
    tick_label.set_fontname("Times New Roman")
# 颜色条标签字体
cbar.set_label(
    "Correlation Coefficient", 
    fontname="Times New Roman", 
    fontsize=14, 
    fontweight="bold"
)

# 8. 调整布局并显示
plt.tight_layout()
plt.show()
```
