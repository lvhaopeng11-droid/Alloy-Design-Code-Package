```python
import pandas as pd
import itertools
column_data = [
    [750],#温度
    [1050,1060,1070,1080,1090,1100],#固溶温度
    [1],#固溶时间
    [700,710,720,730,740,750],#稳定化温度
    [4],#稳定化时间
    [700,710,720,730,740,750],#时效温度
    [1,8,16],#时效时间
]
combinations = list(itertools.product(*column_data))
df = pd.DataFrame(combinations)
df.to_excel('新的实验条件.xlsx', index=False)
```
